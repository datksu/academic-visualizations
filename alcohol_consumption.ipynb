```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
plt.rcParams["font.family"] = "Times New Roman"
```

```python
df = pd.read_csv('vars.csv', sep=';')
df['drink_age'] = df['drink_age'].astype('float')
plt.rcParams["font.family"] = "Times New Roman"
plt.figure(figsize=(10, 6), dpi=500)
sns.kdeplot(data=df, fill=True, alpha=.8, linewidth=0, palette='gray')
plt.title('Distribution of the Respondent Age and the Age of the First Alcohol Use',fontsize=14, fontweight='bold')
plt.legend(title='Variables:', labels=['Age of First Alcohol Use', 'Respondent Age'])
plt.ylabel('Density', labelpad=10, fontsize=12)
plt.xlabel('Age', labelpad=10, fontsize=12)
plt.axvline(29, c = 'black', ls = '--', alpha = .2);
```

```python
df1 = pd.read_csv('total.csv', sep=';')
df1['total'] = df1['total'].astype('float')
plt.rcParams["font.family"] = "Times New Roman"
plt.figure(figsize=(10, 6), dpi=500)
sns.kdeplot(data=df1, fill=True, alpha=.8, linewidth=0, palette='gray')
plt.title('Distribution of the Total Amount of Alcohol Consumed on Average per Occasion',fontsize=14, fontweight='bold')
plt.legend(title='Variable:', labels=['Total amount of Alcohol Consumed on Average per Occasion'])
plt.ylabel('Density', labelpad=10, fontsize=12)
plt.xlabel('Amount (in standard drinks)', labelpad=10, fontsize=12)
plt.axvline(18, c = 'black', ls = '--', alpha = .2);
```

```python
plt.rcParams["font.family"] = "sans-serif"
df2 = pd.read_csv('practice.csv', sep=';')
plt.rcParams["font.family"] = "Times New Roman"
plt.figure(figsize=(10, 6), dpi=500)
ax = sns.countplot(
    x=df2['practice'],
    linewidth=0.5,
    palette=['#d9d9d9', '#bfbfbf', '#a6a6a6'],
    alpha=.8
)
ax.set_xticks(ax.get_xticks())
ax.set_xticklabels([
    'Alcohol Abuse',
    'High Risk',
    'Low Risk'
])
plt.title(
    'Distribution of Respondents by Alcohol Consumption Practices',
    fontsize=14,
    fontweight='bold'
)
plt.ylabel(
    'Number of Respondents',
    labelpad=10,
    fontsize=12
)
plt.xlabel(
    'Group',
    labelpad=10,
    fontsize=12,
    fontweight='bold'
)
plt.show()
```
